```python
from pathlib import Path

# Move up from Model to the project folder, then locate the dataset.
data_dir = Path.cwd().parent / "Dataset_32x32"

# Locate the training and test folders inside the dataset.
train_dir = data_dir / "Training"
test_dir = data_dir / "Test"

print("Dataset found:", data_dir.is_dir())
print("Training folder found:", train_dir.is_dir())
print("Test folder found:", test_dir.is_dir())
```

```python
# Each training subfolder represents a class the model will learn.
class_names = sorted(
    folder.name for folder in train_dir.iterdir()
    if folder.is_dir()
)

# Count supported image files inside each dataset folder.
image_extensions = {".jpg", ".jpeg", ".png"}

def count_images(folder):
    return sum(
        file.is_file() and file.suffix.lower() in image_extensions
        for file in folder.rglob("*")
    )

print("Number of classes:", len(class_names))
print("Training images:", count_images(train_dir))
print("Test images:", count_images(test_dir))
print("First 10 classes:", class_names[:10])

# Training and test folders must contain the same class labels.
test_classes = sorted(
    folder.name for folder in test_dir.iterdir()
    if folder.is_dir()
)
print("Training/test classes match:", class_names == test_classes)
```
